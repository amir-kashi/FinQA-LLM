## Cleaning

```python
# Read the json data (list)
with open("../local_data/train.json") as f:
    data = json.load(f)

data[0].keys()
```

```python
# Create a list to store the extracted data
extracted_data = []

# Iterate over the data
for item in data:
    
    # Check for 'qa', 'qa_0', and 'qa_1' keys and extract question-answer pairs
    for qa_key in ['qa', 'qa_0', 'qa_1']:
        if qa_key in item:
            qa = item[qa_key]
            question = qa.get('question')
            answer = qa.get('answer')
            
            # Append the extracted data to the list
            extracted_data.append({
                'id': item.get('id'),
                'pre_text': item.get('pre_text'),
                'post_text': item.get('post_text'),
                'filename': item.get('filename'),
                'table_ori': item.get('table_ori'),
                'table': item.get('table'),
                'annotation': item.get('annotation'),
                'question': question,
                'answer': answer
            })

# Convert the list to a DataFrame
df_cleaned = pd.DataFrame(extracted_data)
```

```python
# Save the DataFrame to a JSON file
df_cleaned.to_json('../data/extracted_data.json', orient='records', indent=4)
```
